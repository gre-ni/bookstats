### Filtering by language

Initially by English, will be adjustable in utils module.

```python
from bookstats.config import RAW_DATA, FILTERED_DATA
from bookstats.formatting import clean_headers
import numpy as np
import pandas as pd

df = pd.read_csv(RAW_DATA, on_bad_lines='warn')
clean_headers(df)
```

Note: Skipping 4 lines -> not high impact, non-English.

```python
df["language_code"].value_counts()
```

Languages I want to include: en-US, eng, en-GB, en-CA

```python
languages_supported = ["en-US", "eng", "en-GB", "en-CA"]

df_english = df[df["language_code"].isin(languages_supported)]
df_english["language_code"].value_counts()
```

### 'NOT A BOOK' Authors


As per the Goodreads Librarian Manual:
The NAB profile (Not a Book) was created to segregate those items that have ISBNs and ASINs but are not considered books.

```python
df[df["authors"] == "NOT A BOOK"].count()
```

```python
df[df["title"].str.contains("not a book", case=False, na=False)].count()
```

### Book sets and collections

Implementing grain logic of 1 row = 1 edition, so that aggregates aren't skewed by high page count or joint rating.

```python
book_sets = df[df["title"].str.contains(r"#\d-\d", regex=True)]

print(book_sets[["title", "num_pages"]].to_string())
```

Test filtering:

```python
non_bookset_mask = ~df["title"].str.contains(r"#\d-\d", regex=True)

bookset_filtered = df[non_bookset_mask]
```

```python
print(book_sets.count())

difference = df.count() - bookset_filtered.count()
print(difference)
```
