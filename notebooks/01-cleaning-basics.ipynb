### Multiple space characters:

Publishers:

```python
# Take the list of names, return only if it matches the regex to see what kind of results I get:

mask_pub_spaces = raw_data["publisher"].str.contains(r"^.+ {2,}.+$", na=False)
raw_data[mask_pub_spaces][["publisher"]].count()
```

**Overview table**

```python
mask_pub_spaces = raw_data["publisher"].str.contains(r"^.+ {2,}.+$", na=False)
mask_authors_spaces = raw_data["authors"].str.contains(r"^.+ {2,}.+$", na=False)
mask_title_spaces = raw_data["authors"].str.contains(r"^.+ {2,}.+$", na=False)

results = {
    "publisher": raw_data[mask_pub_spaces]["publisher"].nunique(),
    "authors": raw_data[mask_authors_spaces]["authors"].nunique(),
    "title": raw_data[mask_title_spaces]["title"].nunique()
}

pd.Series(results)
```

### Examining before and after


```python
raw_data["publisher_clean"] = raw_data["publisher"].str.replace(r" {2,}"," ",regex=True)

raw_data[["publisher", "publisher_clean"]].head(100)
```
